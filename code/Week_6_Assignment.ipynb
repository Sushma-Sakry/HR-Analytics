```python
# Week 6 Assignment: Git, GitHub & Final Data Analytics Project
# Task 3: Dataset Selection & Data Preparation
import pandas as pd

df = pd.read_csv("HR_Analytics.csv")

print("\nDataset shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values per column:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nExact duplicate rows:", df.duplicated().sum())
```

```python
(df[df.duplicated(keep=False)].sort_values("EmpID"))
```

```python
print("Repeated employee IDs:")
print(df[df["EmpID"].duplicated(keep=False)]["EmpID"].value_counts())
```

```python
df[df["YearsWithCurrManager"].isnull()]
```

```python
print("Columns with only one unique value:")
for column in df.columns:
    if df[column].nunique(dropna=False) == 1:
        print(column, ":", df[column].unique())
```

```python
df_clean = df.copy()
df_clean = df_clean.drop_duplicates()

print("Missing values before cleaning:")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])

median_years = df_clean["YearsWithCurrManager"].median()
df_clean["YearsWithCurrManager"] = df_clean["YearsWithCurrManager"].fillna(median_years)

df_clean["BusinessTravel"] = df_clean["BusinessTravel"].replace(
    "TravelRarely", "Travel_Rarely"
)

df_clean = df_clean.drop(
    columns=["EmployeeCount", "Over18", "StandardHours"]
)

print("\nOriginal dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])

print("\nExact duplicate rows after cleaning:", df_clean.duplicated().sum())

print("\nTravel categories after cleaning:")
print(df_clean["BusinessTravel"].value_counts())

display(df_clean.head())
```

```python
df_clean.to_csv("HR_Analytics_Cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
print("Final rows:", df_clean.shape[0])
print("Final columns:", df_clean.shape[1])
```

```python
df_clean["AttritionFlag"] = df_clean["Attrition"].map(
    {"No": 0, "Yes": 1}
)

correlation = df_clean.corr(numeric_only=True)["AttritionFlag"]
correlation = correlation.drop("AttritionFlag")
correlation = correlation.sort_values(ascending=False)

display(correlation.head(10).round(3))
```

```python
print("\nMonthly Income Summary:")
display(df_clean["MonthlyIncome"].describe())

Q1 = df_clean["MonthlyIncome"].quantile(0.25)
Q3 = df_clean["MonthlyIncome"].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df_clean[
    (df_clean["MonthlyIncome"] < lower_limit) |
    (df_clean["MonthlyIncome"] > upper_limit)
]

print("\nPotential Monthly Income Outliers:", len(outliers))
```

```python
correlation = df_clean.corr(numeric_only=True)["AttritionFlag"]
correlation = correlation.drop("AttritionFlag")

strongest_correlation = correlation.reindex(
    correlation.abs().sort_values(ascending=False).index
)

print("Ten strongest correlations with attrition:")
display(strongest_correlation.head(10).round(3))
```

## Exploratory Data Analysis (EDA) - Key Findings

The cleaned HR dataset contains 1,473 employee records. A total of 237 employees left the company, resulting in an overall attrition rate of 16.09%.

The analysis found that employees working overtime had a higher attrition rate (30.53%) than employees not working overtime (10.41%). The Sales department had the highest department-level attrition rate at 20.58%, while Sales Representatives had the highest job-role attrition rate at 39.29%. Employees aged 18–25 had an attrition rate of approximately 35.77%.

Correlation analysis showed weak negative relationships between attrition and factors such as total working years (-0.171), job level (-0.169), and years in the current role (-0.160). These results suggest that employee experience, role, compensation, and working conditions are useful areas for further investigation. Correlation alone does not establish causation.

The monthly income analysis identified 114 potential outliers using the interquartile range (IQR) method. These records were retained because an outlier is not necessarily an error and requires further investigation before removal.
